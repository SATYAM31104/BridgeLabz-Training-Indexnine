```python
def plot_vector2d(vector2d, origin=[0,0], **options):
    plt.arrow(origin[0], origin[1],
              vector2d[0], vector2d[1],
              head_width=0.2, head_length=0.3,
              length_includes_head=True, **options)
plot_vector2d(u, color='r')
plot_vector2d(v, color='b')
plt.xlim(0, 6)
plt.ylim(0, 6)
plt.grid()
plt.scatter(u,v)
plt.show()
```

```python
u = np.array([2, 5])
v = np.array([3, 1])
def vector_angle(u, v):
    cos_theta = u.dot(v) / (np.linalg.norm(u) * np.linalg.norm(v))
    return np.arccos(np.clip(cos_theta, -1, 1))
angle_rad = vector_angle(u, v)
angle_deg = np.degrees(angle_rad)
print(f"Angle in radians: {angle_rad}")
print(f"Angle in degrees: {angle_deg}")
```
