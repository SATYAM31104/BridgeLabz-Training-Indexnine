```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
sns.set_theme(style='white')
```

```python
# Load HDFC Dataset
df = pd.read_csv('HDFC_Dataset.csv', header=0)
display(df.head())
print(df.info())
```

```python
# Calculate Mean of Applicant Income
print("Mean Applicant Income:", df['Applicant_Income'].mean())
```

```python
# Descriptive Statistics
display(df.describe())
```

```python
# Group by Customer Name and mean income
display(df.groupby("Customer_Name")['Applicant_Income'].mean().sort_values(ascending=False))
```

```python
# Plot Histogram
# Fix: Use data=df, x='column', hue='column' parameters for correct seaborn handling
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x="Applicant_Income", hue="Loan_Status", kde=True)
plt.title("Applicant Income Distribution by Loan Status")
plt.show()
```

```python
sns.boxplot(x="Loan_Amount", y="Loan_Status", data=df)
plt.show()
```
