```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# Load data (Adjust path if needed)
df = pd.read_csv('HDFC_Dataset.csv')

# --- Q6. Probability: Self-Employed AND Loan > 10,00,000 ---
# P(A and B) = count(A and B) / total_count
event_match = df[(df['Employment_Status'] == 'Self-Employed') & (df['Loan_Amount'] > 1000000)]
prob_q6 = len(event_match) / len(df)
print(f"Q6. Probability (Self-Employed & Loan > 10L): {prob_q6:.4f} ({prob_q6*100:.2f}%)")

# --- Q7. Random Variables: Loan_Amount Distribution ---
plt.figure(figsize=(10, 6))
# Histogram with Density Curve (kde=True)
sns.histplot(df['Loan_Amount'], kde=True, color='blue')
plt.title('Q7. Distribution of Loan Amount')
plt.xlabel('Loan Amount')
plt.show()
print("Q7 Comment: Observe if the curve is bell-shaped (Normal) or skewed to one side.")

# --- Q8. Normality Test (Shapiro-Wilk) on Applicant_Income ---
# H0: Data is Normal. H1: Data is NOT Normal.
stat, p_value = stats.shapiro(df['Applicant_Income'])
print(f"\nQ8. Shapiro-Wilk Test on Applicant_Income:\nStatistic: {stat:.4f}, p-value: {p_value:.10f}")
if p_value > 0.05:
    print("Result: Fail to reject H0 (Data looks Normal)")
else:
    print("Result: Reject H0 (Data is NOT Normal)")

# --- Q9. Conditional Probability: P(Urban | Approved) ---
# P(Urban | Approved) = P(Urban AND Approved) / P(Approved)
approved_df = df[df['Loan_Status'] == 'Approved']
urban_approved = approved_df[approved_df['Property_Area'] == 'Urban']
p_urban_given_approved = len(urban_approved) / len(approved_df)
print(f"\nQ9. P(Urban | Approved): {p_urban_given_approved:.4f} ({p_urban_given_approved*100:.2f}%)")

# --- Q10. Bayes' Theorem: P(Good Credit | Approved) ---
# P(A|B) = P(B|A) * P(A) / P(B)
# A = Good Credit (1), B = Approved
p_good_credit = len(df[df['Credit_History'] == 1]) / len(df)       # P(A)
p_approved = len(df[df['Loan_Status'] == 'Approved']) / len(df)    # P(B)

good_credit_df = df[df['Credit_History'] == 1]
p_approved_given_good_credit = len(good_credit_df[good_credit_df['Loan_Status'] == 'Approved']) / len(good_credit_df) # P(B|A)

bayes_result = (p_approved_given_good_credit * p_good_credit) / p_approved
print(f"\nQ10. Bayes' Theorem Result: {bayes_result:.4f}")

# --- Q11. Covariance Matrix ---
cols = ['Applicant_Income', 'Loan_Amount', 'CIBIL_Score', 'Debt_to_Income_Ratio']
print("\nQ11. Covariance Matrix:")
print(df[cols].cov())

# --- Q12. Pearson Correlation ---
corr1 = df['Loan_Amount'].corr(df['Annual_Household_Income'])
corr2 = df['Loan_Amount'].corr(df['CIBIL_Score'])
print(f"\nQ12. Pearson Correlation:")
print(f"Loan_Amount vs Household_Income: {corr1:.4f}")
print(f"Loan_Amount vs CIBIL_Score:      {corr2:.4f}")

# --- Q13. Hypothesis Testing (T-test) ---
# H0: Mean Loan Amount is same for Graduates and Non-Graduates
grad = df[df['Education'] == 'Graduate']['Loan_Amount']
non_grad = df[df['Education'] == 'Not Graduate']['Loan_Amount']

t_stat, p_val_ttest = stats.ttest_ind(grad, non_grad)
print(f"\nQ13. T-Test Results:\nT-statistic: {t_stat:.4f}, p-value: {p_val_ttest:.4f}")
if p_val_ttest < 0.05:
    print("Result: Significant difference found (Reject H0).")
else:
    print("Result: No significant difference (Fail to reject H0).")
```

```python
#question 13
import pandas as pd
from scipy import stats

# Load data
df = pd.read_csv('HDFC_Dataset.csv')

# 1. Split the data into two groups
group_grad = df[df['Education'] == 'Graduate']['Loan_Amount']
group_non_grad = df[df['Education'] == 'Not Graduate']['Loan_Amount']

# 2. Perform Independent T-test
# equal_var=False performs Welch's T-test (safer assumption that variances aren't identical)
t_stat, p_value = stats.ttest_ind(group_grad, group_non_grad, equal_var=False)

# 3. Output
print(f"Mean Loan (Graduate):     ₹{group_grad.mean():,.2f}")
print(f"Mean Loan (Not Graduate): ₹{group_non_grad.mean():,.2f}")
print("-" * 50)
print(f"T-statistic: {t_stat:.4f}")
print(f"P-value:     {p_value:.5f}")

print("\n--- CONCLUSION ---")
alpha = 0.05
print(f"Significance Level: {alpha}")

if p_value < alpha:
    print("Result: REJECT Null Hypothesis.")
    print("Meaning: There is a SIGNIFICANT difference in loan amounts between Graduates and Non-Graduates.")
else:
    print("Result: FAIL TO REJECT Null Hypothesis.")
    print("Meaning: There is NO statistically significant difference in loan amounts.")
```

```python
import pandas as pd
from scipy.stats import chi2_contingency

# Load data
df = pd.read_csv('HDFC_Dataset.csv')

# 1. Create a Contingency Table (Cross-tabulation)
# This shows the counts for each combination of Gender and Loan_Status
contingency_table = pd.crosstab(df['Gender'], df['Loan_Status'])

print("--- Contingency Table (Counts) ---")
print(contingency_table)
print("-" * 35)

# 2. Perform the Chi-square Test
# chi2: The test statistic
# p: The p-value
# dof: Degrees of freedom
# expected: The counts we would expect if there was NO relationship
chi2, p, dof, expected = chi2_contingency(contingency_table)

print(f"Chi-square Statistic: {chi2:.4f}")
print(f"P-value:             {p:.5f}")

print("\n--- INTERPRETATION ---")
alpha = 0.05

if p < alpha:
    print(f"Result: P-value ({p:.4f}) is LESS than 0.05.")
    print("Conclusion: REJECT the Null Hypothesis.")
    print("-> There IS a significant relationship between Gender and Loan Approval.")
else:
    print(f"Result: P-value ({p:.4f}) is GREATER than 0.05.")
    print("Conclusion: FAIL TO REJECT the Null Hypothesis.")
    print("-> Gender and Loan Approval are INDEPENDENT. (No significant bias found).")
```

```python
import numpy as np
from scipy import stats

# Calculate mean, standard error of the mean (SEM), and interval
loan_data = df['Loan_Amount']
mean_loan = np.mean(loan_data)
sem = stats.sem(loan_data) # Standard Error

# 95% CI using T-distribution
ci_95 = stats.t.interval(0.95, len(loan_data)-1, loc=mean_loan, scale=sem)

print(f"95% Confidence Interval for Loan Amount: ₹{ci_95[0]:,.2f} to ₹{ci_95[1]:,.2f}")
print(f"Interpretation: We are 95% confident the true population average loan is in this range.")
```

```python
dti_data = df['Debt_to_Income_Ratio']
ci_dti = stats.t.interval(0.95, len(dti_data)-1, loc=np.mean(dti_data), scale=stats.sem(dti_data))

print(f"95% CI for DTI Ratio: {ci_dti[0]:.4f} to {ci_dti[1]:.4f}")
print("Importance: Helps banks assess if their 'average' customer is over-leveraged.")
```
